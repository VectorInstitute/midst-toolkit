```python
FIG_HEIGHT = 800
FIG_WIDTH = 1100
FONT_SIZE = 32
```

## Utilities

```python
repo_abs_path = Path(os.path.abspath("")).parent.parent

plots_dir = f"{repo_abs_path}/examples/visualizations/diabetes_ensemble_attacker/"
```

```python
def customize_figure_layout(gen_fig: Any, render_legend: bool = True) -> None:
    if render_legend:
        gen_fig.update_layout(
            height=FIG_HEIGHT,
            width=FIG_WIDTH,
            font_color="black",
            legend=dict(
                y=1.0,
                x=0.5,
                xanchor="center",
                yanchor="bottom",
                orientation="h",
                valign="top",
                title_text="",
                font=dict(size=FONT_SIZE+4),
                title_font_family="Helvetica",
            ),
            title_font_family="Helvetica",
            title_x=0.5,
            title_y=0.975,
            margin=dict(l=30, r=30, t=75, b=25),
            plot_bgcolor="white",
            font=dict(size=FONT_SIZE+10, family="Helvetica"),
        )
    else:
        gen_fig.update_layout(
            height=FIG_HEIGHT,
            width=FIG_WIDTH,
            font_color="black",
            legend={},
            title_font_family="Helvetica",
            title_x=0.5,
            title_y=0.975,
            margin=dict(l=30, r=30, t=75, b=25),
            plot_bgcolor="white",
            font=dict(size=FONT_SIZE + 10, family="Helvetica"),
        )


def save_and_display_figure(gen_fig: Any, template_name: str, plots_dir: str) -> None:
    plot_png_path = f"{plots_dir}/{template_name}.png"
    plot_pdf_path = f"{plots_dir}/{template_name}.pdf"

    os.makedirs(plots_dir, exist_ok=True)
    gen_fig.write_image(plot_png_path, scale=2)
    gen_fig.write_image(plot_pdf_path)

    display(gen_fig)
```

```python
df = {"Ensemble Variations": ensemble_variations, "mia_ensemble_ablation": mia_ensemble_ablation}

fig = px.bar(
    data_frame=df,
    x="Ensemble Variations",
    y="mia_ensemble_ablation",
)

fig.update_traces(
    marker=dict(color=["#53abff", "#f78d8d", "#f78d8d", "#f78d8d", "#660066", "#660066", "#660066"])
)

fig.update_layout(bargap=0.0, showlegend=False)
fig.update_xaxes(
  ticks="outside", tickwidth=4, ticklen=7.5, linewidth=4, linecolor="black", tickangle=0, title_font_family="Helvetica", tickfont_family="Helvetica", tickfont_size=FONT_SIZE,
)
fig.update_yaxes(
    ticks="outside", tickwidth=4, ticklen=7.5, linewidth=4, linecolor="black", title = "Ensemble (BB) Success", tickangle=0, title_font_family="Helvetica", tickfont_family="Helvetica", tickfont_size=FONT_SIZE+10,
)

fig.update_yaxes(tickmode='array', tickvals=[0.1, 0.2, 0.3, 0.4], range=[0, 0.4])
fig.update_traces(marker_line_width=4, marker_line_color="black")
fig.update_xaxes(title_text="Diabetes Ensemble Ablations")
fig.update_traces(showlegend=False)
fig.update(layout_coloraxis_showscale=False)

customize_figure_layout(fig, render_legend=False)
save_and_display_figure(fig, "diabetes_ensemble_ablation", plots_dir)
```

```python
mismatched_specifics = ["Less", "Default", "More"]
mismatch_type = ["Train Time", "Train Time", "Train Time"]
mismatch_mia = [0.11733, 0.281, 0.271333]
df_1 = {"mismatched_specifics": mismatched_specifics, "Mismatch Type": mismatch_type, "mismatch_mia": mismatch_mia}

mismatched_specifics_2 = ["Least", "Less", "Default"]
mismatch_type_2 = ["Diffusion Steps", "Diffusion Steps", "Diffusion Steps"]
mismatch_mia_2 = [0.2806667, 0.26933, 0.281]
df_2 = {"mismatched_specifics": mismatched_specifics_2, "Mismatch Type": mismatch_type_2, "mismatch_mia": mismatch_mia_2}

mismatched_specifics_3 = ["Small", "Default", "Large"]
mismatch_type_3 = ["Model Size", "Model Size", "Model Size"]
mismatch_mia_3 = [0.132, 0.281, 0.28733333]
df_3 = {"mismatched_specifics": mismatched_specifics_3, "Mismatch Type": mismatch_type_3, "mismatch_mia": mismatch_mia_3}

fig = go.Figure()

fig1 = px.bar(
    data_frame=df_1,
    x="mismatched_specifics",
    y="mismatch_mia",
    color="mismatched_specifics",
    color_discrete_sequence=["#f78d8d", "#53abff", "#660066"],
)

fig2 = px.bar(
    data_frame=df_2,
    x="mismatched_specifics",
    y="mismatch_mia",
    color="mismatched_specifics",
    color_discrete_sequence=["#f78d8d", "#660066", "#53abff"],
)

fig3 = px.bar(
    data_frame=df_3,
    x="mismatched_specifics",
    y="mismatch_mia",
    color="mismatched_specifics",
    color_discrete_sequence=["#f78d8d", "#53abff", "#660066"],
)

fig = make_subplots(rows=1, cols=3, shared_yaxes=True, horizontal_spacing=0.08)

for trace in fig1.data:
    fig.add_trace(trace, row=1, col=1)

for trace in fig2.data:
    fig.add_trace(trace, row=1, col=2)

for trace in fig3.data:
    fig.add_trace(trace, row=1, col=3)

fig.update_layout(bargap=0.0, showlegend=False)
fig.update_xaxes(
  ticks="outside", tickwidth=4, ticklen=7.5, linewidth=4, linecolor="black", tickangle=-35, title_font_family="Helvetica", tickfont_family="Helvetica", tickfont_size=FONT_SIZE+10,
)
fig.update_yaxes(
    ticks="outside", tickwidth=4, ticklen=7.5, linewidth=4, linecolor="black", title = "Ensemble (BB) Success", tickangle=0, title_font_family="Helvetica", tickfont_family="Helvetica", tickfont_size=FONT_SIZE+10,
)
fig.update_yaxes(tickmode='array', tickvals=[0.1, 0.2, 0.3], range=[0, 0.3])
fig.update_traces(marker_line_width=4, marker_line_color="black")
fig.update_yaxes(showticklabels=False, title_text=None, row=1, col=2)
fig.update_xaxes(title_text="Train Time", categoryorder="array", categoryarray=["Less", "Default", "More"], row=1, col=1)
fig.update_xaxes(title_text="Diffusion Steps", categoryorder="array", categoryarray=["Least", "Less", "Default"], row=1, col=2)
fig.update_xaxes(title_text="Model Size", categoryorder="array", categoryarray=["Small", "Default", "Large"], row=1, col=3)
fig.update_yaxes(visible=False, row=1, col=2)
fig.update_yaxes(visible=False, row=1, col=3)

customize_figure_layout(fig, render_legend=False)
save_and_display_figure(fig, "diabetes_ensemble_shadow_variation", plots_dir)
```
